```python
import pickle
import pandas as pd

with open('arrest_model.pkl', 'rb') as f:
    model = pickle.load(f)
```

```python
stops = pd.read_parquet('open_policing_project.parquet').sample(100)    # or your own records in the same schema
predictions = model.predict(stops)  
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(stops['is_arrested'], predictions['predicted_arrest'])
```
